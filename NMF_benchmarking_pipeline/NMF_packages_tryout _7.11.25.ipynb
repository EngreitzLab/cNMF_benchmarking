## sklearn

```python
import numpy as np
from sklearn.decomposition import NMF as SKNMF

X = np.array([[1, 2], [2, 2], [3, 1.2], [4, 1], [5, 0.8], [6, 1]])
model = SKNMF(n_components=2, init='random', random_state=0)
W = model.fit_transform(X)
H = model.components_
```

```python
print("X (Orginal matrix):")
print(X)
print("\nW (basis matrix):")
print(W)
print("\nH (coefficient matrix):")
print(H)    
X_hat = np.dot(W, H)
print("\nX_hat (Reconstructed matrix):")
print(X_hat)
```

## torchnmf

```python
import torch
from torchnmf.nmf import NMF as TNMF

# Convert to torch tensor
X_torch = torch.from_numpy(X).float()

# Initialize the NMF model with desired rank (number of components)
rank = 2
model = TNMF(X_torch.shape, rank=rank)
```

```python
print("X (Orginal matrix):")
print(X_torch)
print("\nW (basis matrix):")
print(model.W)
print("\nH (coefficient matrix):")
print(model.H)    
X_hat = np.dot(W, H)
print("\nX_hat (Reconstructed matrix):")
print(X_hat)
```
